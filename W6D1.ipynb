```python

import os
from groq import Groq

# Read the API key from an environment variable (never hardcode it).
api_key = os.environ.get("GROQ_API_KEY")

if not api_key:
    raise ValueError(
        "GROQ_API_KEY is missing. Set it in your terminal first."
    )

client = Groq(api_key=api_key)

# Use the same question in both experiments.
question = "What is an API? Explain it using a simple everyday example."

experiments = [
    (
        "Patient Teacher",
        "You are a patient teacher. Explain the concept step by "
        "step using simple language and an everyday example."
    ),
    (
        "Concise Expert",
        "You are a concise expert. Answer accurately in no more "
        "than three short sentences."
    ),
]

responses = {}

for name, system_prompt in experiments:
    print("\n" + "=" * 60)
    print(f"Experiment: {name}")
    print(f"System prompt: {system_prompt}")
    print(f"User question: {question}")
    print("=" * 60)

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question},
        ],
    )

    answer = response.choices[0].message.content
    responses[name] = answer

    # Print the model's response.
    print("\nModel response:")
    print(answer)

    # Read token counts directly from the API response.
    usage = response.usage
    print("\nToken usage from API response:")
    print(f"Prompt tokens: {usage.prompt_tokens}")
    print(f"Completion tokens: {usage.completion_tokens}")
    print(f"Total tokens: {usage.total_tokens}")

# Compare the two responses.
print("\n" + "=" * 60)
print("COMPARISON")
print("=" * 60)
print("\nPatient Teacher response:")
print(responses["Patient Teacher"])
print("\nConcise Expert response:")
print(responses["Concise Expert"])

print("\nThe first prompt requests a detailed explanation,")
print("while the second requests a maximum of three short sentences.")
print("Compare their detail, clarity, and completion token counts.")
```
