```python

from sentence_transformers import SentenceTransformer, util
from itertools import combinations

# Load the model
model = SentenceTransformer("all-MiniLM-L6-v2")

# 10 sentences about football, programming, and food
sentences = [
    "The football team won the championship.",
    "The players celebrated after winning the league.",
    "A striker scored three goals in the match.",
    "Python is used for data analysis.",
    "The programmer fixed a bug in the code.",
    "Coding helps people automate daily tasks.",
    "The chef cooked a delicious meal.",
    "She prepared a tasty dinner with fresh vegetables.",
    "He deposited his money at the bank.",
    "The fisherman sat on the bank of the river.",
]

# My predictions before checking the results
most_similar_guess = [(1, 2), (7, 8), (4, 6)]
least_similar_guess = [(3, 10), (4, 7), (5, 9)]

print("My predictions:")
print("Most similar pairs:", most_similar_guess)
print("Least similar pairs:", least_similar_guess)

# Turn the sentences into embeddings
embeddings = model.encode(sentences, convert_to_tensor=True)

print("\nNumber of sentences:", len(sentences))
print("Embedding dimensions:", embeddings.shape[1])
print("First five values:", embeddings[0][:5].tolist())

# Compare every possible pair
results = []

for (i, first), (j, second) in combinations(enumerate(sentences), 2):
    score = util.cos_sim(embeddings[i], embeddings[j]).item()
    results.append((score, i + 1, j + 1, first, second))

results.sort(reverse=True)

print("\nThree most similar pairs:")
for score, i, j, first, second in results[:3]:
    print(f"\nPair {i} and {j}: {score:.4f}")
    print(first)
    print(second)

print("\nThree least similar pairs:")
for score, i, j, first, second in results[-3:]:
    print(f"\nPair {i} and {j}: {score:.4f}")
    print(first)
    print(second)

# Check the two different meanings of "bank"
bank_result = next(
    item for item in results
    if {item[1], item[2]} == {9, 10}
)

print("\nThe two meanings of bank:")
print(sentences[8])
print(sentences[9])
print("Similarity:", round(bank_result[0], 4))

# Bonus: try four Arabic sentences
arabic_sentences = [
    "فاز الفريق بالمباراة.",
    "حقق اللاعبون الفوز في اللقاء.",
    "تستخدم بايثون في تحليل البيانات.",
    "البرمجة تساعد على حل المشكلات.",
]

arabic_embeddings = model.encode(
    arabic_sentences, convert_to_tensor=True
)

print("\nArabic sentence comparisons:")

for (i, first), (j, second) in combinations(
    enumerate(arabic_sentences), 2
):
    score = util.cos_sim(
        arabic_embeddings[i], arabic_embeddings[j]
    ).item()

    print(f"\nPair {i + 1} and {j + 1}: {score:.4f}")
    print(first)
    print(second)

# Save the results in notes.md
with open("notes.md", "a", encoding="utf-8") as file:
    file.write("\n\n# W6D3 - Embeddings\n")

    file.write("\n## What is an embedding?\n")
    file.write(
        "\nAn embedding represents a sentence as a list of numbers. "
        "These numbers help us compare how similar two sentences "
        "are in meaning.\n"
    )

    file.write("\n## My predictions\n")
    file.write(f"\nMost similar pairs I guessed: {most_similar_guess}\n")
    file.write(f"\nLeast similar pairs I guessed: {least_similar_guess}\n")

    file.write("\n## English results\n")
    file.write("\n| Sentence 1 | Sentence 2 | Similarity |\n")
    file.write("|---|---|---:|\n")

    for score, i, j, first, second in sorted(
        results, key=lambda item: (item[1], item[2])
    ):
        file.write(
            f"| {i}: {first} | {j}: {second} | {score:.4f} |\n"
        )

    file.write("\n## Most similar pairs\n")
    for score, i, j, first, second in results[:3]:
        file.write(f"\n- Pair ({i}, {j}): {score:.4f}\n")
        file.write(f"  - {first}\n  - {second}\n")

    file.write("\n## Least similar pairs\n")
    for score, i, j, first, second in results[-3:]:
        file.write(f"\n- Pair ({i}, {j}): {score:.4f}\n")
        file.write(f"  - {first}\n  - {second}\n")

    file.write("\n## The word bank\n")
    file.write(
        "\nThe two sentences use the word 'bank' in different ways. "
        "I will compare their score to see how well the model uses context.\n"
    )

    file.write("\n## Arabic bonus\n")
    file.write(
        "\nI compared four Arabic sentences as well. Since this model "
        "is mainly designed for English, its Arabic results may not "
        "be as reliable.\n"
    )

    file.write("\n## My observations\n")
    file.write(
        "\n- Were my predictions correct?\n"
        "- What surprised me about the results?\n"
        "- How similar were the two meanings of 'bank'?\n"
        "- Did the model handle Arabic well?\n"
    )

print("\nDone! Check the results and notes.md.")
```
